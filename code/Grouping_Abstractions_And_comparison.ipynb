```python
abstractions = []
used_words = set()  # fast lookup instead of scanning abstractions

for group in globaldict:
    for word in globaldict[group]:

        # Skip if already assigned to a concept
        if word in used_words:
            continue

        superset = set()

        # First-level neighbors
        first_neighbors = find_word(word, loaded_dict)

        for fn in first_neighbors:

            if fn in used_words:
                continue

            # Second-level neighbors
            second_neighbors = find_word(fn, loaded_dict)

            for sn in second_neighbors:
                if sn not in used_words:
                    superset.add(sn)

        # Also include first neighbors + the original word
        first_neighbors = set([fn for fn in first_neighbors if fn not in used_words])
        if first_neighbors:
            superset.update(first_neighbors)
        
            superset.add(word)

            # Mark all words as used
            used_words.update(superset)

        abstractions.append(superset)
```

```python
def merge_abstractions_unique(abstractions, overlap_threshold=2):
    # make a copy of sets
    sets = [set(s) for s in abstractions]
    clusters = []

    while sets:
        # pick the first remaining set
        current = sets.pop(0)
        
        merged = True
        while merged:
            merged = False
            for s in sets[:]:  # iterate over a copy
                # merge if overlap
                if len(current & s) >= overlap_threshold:
                    current.update(s)
                    sets.remove(s)
                    merged = True
        clusters.append(current)
        
        # remove these words from remaining sets
        for s in sets:
            s -= current
        # remove empty sets
        sets = [s for s in sets if s]

    return clusters
merge=merge_abstractions_unique(globaldict.values(), overlap_threshold=4)
```

```python
def merge_groups(groups, min_overlap=3, hub_thresh=200):
    from collections import defaultdict
    
    word_to_groups = defaultdict(set)
    for i, g in enumerate(groups):
        for w in g:
            word_to_groups[w].add(i)

    merged = []
    visited = set()

    for i in range(len(groups)):
        if i in visited:
            continue

        cluster = set(groups[i])
        stack = [i]
        visited.add(i)

        while stack:
            g_idx = stack.pop()
            for w in groups[g_idx]:
                if len(word_to_groups[w]) > hub_thresh:
                    continue
                
                for j in word_to_groups[w]:
                    if j in visited:
                        continue
                    
                    if len(cluster & groups[j])/len(cluster | groups[j]) > 0.25: # min_overlap:
                        cluster |= groups[j]
                        visited.add(j)
                        stack.append(j)

        merged.append(cluster)

    return merged
merged_clusters=merge_groups(list(globaldict.values()))
```
