```python
#preserved concepts
import concept_getters
import pandas as pd
import re
SPARSITY_RATIOS = [20.0, 36.0, 48.8, 59.04, 67.232, 73.786, 79.028, 83.223, 86.578, 89.263, 91.41, 93.128, 94.502]
explanations = {}

def get_indiv_concepts_per_unit(df) -> dict:
    concepts = {}
    for unit,expl in zip(df.unit, df.best_name):
        concps = re.findall(r'(?<!\bNOT\s)(?:\b(?:hyp|pre|oth):[^\s)]+)', expl)
        for i in range(len(concps)):
            while concps[i][-1] == ')':
                concps[i] = concps[i][:-1]
        concepts[unit] = concps
    return concepts

for sparsity in SPARSITY_RATIOS:
    cluster3_expls = pd.read_csv(f"/workspace/CCE_NLI/BERT/exp/lottery_ticket/Run2Full_allneurons/Expls/{sparsity}_Pruning_Iter/Cluster3IOUS1024N.csv")
    explanations[sparsity] = get_indiv_concepts_per_unit(cluster3_expls) 
```
